````python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# 🔧 FUNCIONES
def load_scaled_data(target=TARGET):
    # 1. Cargar features seleccionados
    selected_features = open(SELECTED_FEATURES_FILE).read().splitlines()
    
    exclude_features = [TARGET, "quarter", "region_count", "price_deviation_from_median", "time_trend", "region_target_encoded", "region_count"]
    
    selected_features = [f for f in selected_features if f not in exclude_features]

    # 2. Cargar datasets
    train = pd.read_parquet(TRAIN_DATA_FILE)
    test = pd.read_parquet(TEST_DATA_FILE)

    # 3. Separar X e y
    
    
    X_train_raw = train[selected_features]
    X_test_raw = test[selected_features]
    
    y_train = train[target].reset_index(drop=True)
    y_test = test[target].reset_index(drop=True)

    # 4. Escalar X con StandardScaler
    scaler = StandardScaler().fit(X_train_raw)
    X_train = pd.DataFrame(scaler.transform(X_train_raw), columns=selected_features)
    X_test = pd.DataFrame(scaler.transform(X_test_raw), columns=selected_features)
    
    if TARGET in X_train.columns:
        X_train = X_train.drop(columns=[TARGET])
        X_test = X_test.drop(columns=[TARGET])


    return X_train, X_test, y_train, y_test, selected_features, scaler
````

````python
# 📦 CARGAR DATA
X_train, X_test, y_train, y_test, features, scaler = load_scaled_data(target=TARGET)
````

````python
import xgboost as xgb
from sklearn.metrics import mean_squared_error
import numpy as np

xgb_model = xgb.XGBRegressor(
    n_estimators=300,
    max_depth=8,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    tree_method="hist",   # optimizado para CPU
    random_state=42,
    n_jobs=-1
)
xgb_model.fit(X_train, y_train)
xgb_preds = xgb_model.predict(X_test)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_preds))
print("XGBoost (CPU) RMSE:", xgb_rmse)
````

````python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=400,
    max_depth=None,
    min_samples_leaf=2,
    n_jobs=-1,
    random_state=42
)
rf_model.fit(X_train, y_train)
rf_preds = rf_model.predict(X_test)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_preds))
print("RandomForest RMSE:", rf_rmse)
````

````python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

ridge = Ridge(alpha=10.0, random_state=42)
ridge.fit(X_train, y_train)
ridge_rmse = np.sqrt(mean_squared_error(y_test, ridge.predict(X_test)))
print("Ridge RMSE:", ridge_rmse)

lasso = Lasso(alpha=0.001, random_state=42, max_iter=20000)
lasso.fit(X_train, y_train)
lasso_rmse = np.sqrt(mean_squared_error(y_test, lasso.predict(X_test)))
print("Lasso RMSE:", lasso_rmse)

enet = ElasticNet(alpha=0.001, l1_ratio=0.5, random_state=42, max_iter=20000)
enet.fit(X_train, y_train)
enet_rmse = np.sqrt(mean_squared_error(y_test, enet.predict(X_test)))
print("ElasticNet RMSE:", enet_rmse)
````

````python
# Compatibilidad: algunos releases de flaml intentan importar
# _is_pandas_na desde sklearn.utils (que no existe en sklearn moderno).
# Creamos el atributo en sklearn.utils antes de importar flaml.
import sklearn
import sklearn.utils
import pandas as pd


from flaml import AutoML

automl = AutoML()
automl.fit(
    X_train=X_train, y_train=y_train,
    task="regression",
    time_budget=600,        # segundos
    metric="rmse",
    estimator_list=["rf", "xgboost"],  # <- nombre correcto
    seed=42,
)

flaml_preds = automl.predict(X_test)
flaml_rmse = np.sqrt(mean_squared_error(y_test, flaml_preds))
print("FLAML AutoML RMSE:", flaml_rmse)
print("Mejor estimador:", automl.best_estimator)
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

def calculate_mape(y_true, y_pred):
    mask = y_true != 0
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

def eval_array_preds(preds, y_true, name):
    rmse = np.sqrt(mean_squared_error(y_true, preds))
    mae = mean_absolute_error(y_true, preds)
    r2 = r2_score(y_true, preds)
    mape = calculate_mape(y_true.values, preds)
    return {'Modelo': name, 'RMSE': rmse, 'MAE': mae, 'R²': r2, 'MAPE (%)': mape,
            'Predicciones': preds, 'Reales': y_true.values}

# Construir resultados (con lo que ya entrenaste)
results = []
results.append(eval_array_preds(xgb_preds, y_test, "XGBoost (CPU)"))

results.append(eval_array_preds(rf_preds, y_test, "RandomForest"))

results.append(eval_array_preds(ridge.predict(X_test), y_test, "Ridge"))
results.append(eval_array_preds(lasso.predict(X_test), y_test, "Lasso"))
results.append(eval_array_preds(enet.predict(X_test), y_test, "ElasticNet"))


# Si FLAML corrió:
results.append(eval_array_preds(flaml_preds, y_test, "FLAML AutoML"))

comparison_df = pd.DataFrame(results)[['Modelo','RMSE','MAE','R²','MAPE (%)']].round(4)
print(comparison_df.to_string(index=False))

best_idx = comparison_df['RMSE'].idxmin()
best_model_name = comparison_df.iloc[best_idx]['Modelo']
print(f"\n🥇 Mejor modelo: {best_model_name}")
````

````python
import matplotlib.pyplot as plt
import numpy as np

def plot_feature_importance(model, feature_names, title):
    importances = model.feature_importances_
    idx = np.argsort(importances)[::-1][:20]
    plt.figure(figsize=(10,6))
    plt.barh(range(len(idx)), importances[idx][::-1])
    plt.yticks(range(len(idx)), np.array(feature_names)[idx][::-1])
    plt.xlabel('Importancia')
    plt.title(title)
    plt.tight_layout()
    plt.show()

plot_feature_importance(xgb_model, X_train.columns, "XGBoost Feature Importance (Top 20)")
plot_feature_importance(rf_model, X_train.columns, "RandomForest Feature Importance (Top 20)")
````
